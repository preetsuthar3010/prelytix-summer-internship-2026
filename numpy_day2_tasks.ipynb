## Task 1.4.1

```
import numpy as np

matrix = np.random.randint(1, 101, (6, 6))

result = matrix[(matrix % 3 == 0) & (matrix % 5 == 0)]

print("Matrix:\n", matrix)
print("Divisible by both 3 and 5:\n", result)
```

## Task 1.4.2

```
import numpy as np

arr = np.random.randint(1, 100, 10)

mean_val = arr.mean()

binary = np.where(arr < mean_val, 0, 1)

print("Array:", arr)
print("Mean:", mean_val)
print("Binarised:", binary)
```

## Task 1.4.3

```
import numpy as np

arr = np.arange(25).reshape(5, 5)

rows = [1, 3, 4]
cols = [0, 2, 4]

sub_matrix = arr[np.ix_(rows, cols)]

print("Original Matrix:\n", arr)
print("Sub Matrix:\n", sub_matrix)
```

## Task 1.4.4

```
import numpy as np

matrix = np.random.randint(1, 100, (4, 4))

row_sorted = np.sort(matrix, axis=1)

col_sorted = np.sort(matrix, axis=0)

print("Original Matrix:\n", matrix)
print("Row-wise Sorted:\n", row_sorted)
print("Column-wise Sorted:\n", col_sorted)
```

## Task 1.4.5

```
import numpy as np

scores = np.random.randint(0, 101, 15)

conditions = [
    scores >= 85,
    scores >= 70,
    scores >= 55,
    scores >= 40
]

grades = ['A', 'B', 'C', 'D']

result = np.select(conditions, grades, default='F')

print("Scores:", scores)
print("Grades:", result)
```

## Task 2.2.1

```
import numpy as np

matrix = np.random.randint(1, 20, (5, 4))

row_add = np.array([1, 2, 3, 4])

col_sub = np.array([10, 20, 30, 40, 50]).reshape(5, 1)

result = matrix + row_add - col_sub

print("Original Matrix:\n", matrix)
print("Result:\n", result)
```

## Task 2.2.2

```
import numpy as np

matrix = np.random.randint(1, 100, (6, 4)).astype(float)

mean = matrix.mean(axis=0)

std = matrix.std(axis=0)

normalized = (matrix - mean) / std

print("Original Matrix:\n", matrix)
print("Normalized Matrix:\n", normalized)
```

## Task 2.2.3

```
import numpy as np

points = np.array([2, 5, 8, 12, 20])

distance_matrix = np.abs(points[:, np.newaxis] - points)

print("Points:", points)
print("Distance Matrix:\n", distance_matrix)
```

## Task 3.3.1

```
import numpy as np

A = np.random.rand(3, 3)

A_inv = np.linalg.inv(A)

identity = A @ A_inv

print("A:\n", A)
print("A @ inv(A):\n", identity)

print("Is Close to Identity?",
      np.allclose(identity, np.eye(3)))
```

## Task 3.3.2

```
import numpy as np

A = np.array([[2, 1],
              [1, 3]])

b = np.array([5, 10])

solution = np.linalg.solve(A, b)

print("Solution:", solution)

print("Verification:", A @ solution)
```

## Task 3.3.3

```
import numpy as np

A = np.random.rand(3, 3)

S = A.T @ A

eigenvalues, eigenvectors = np.linalg.eig(S)

print("Symmetric Matrix:\n", S)

print("Eigenvalues:\n", eigenvalues)

print("All Non-negative?",
      np.all(eigenvalues >= 0))
```

## Task 3.3.4

```
import numpy as np

v1 = np.random.rand(5)

v2 = np.random.rand(5)

u1 = v1 / np.linalg.norm(v1)

u2 = v2 / np.linalg.norm(v2)

dot_product = np.dot(u1, u2)

print("Unit Vector 1:", u1)
print("Unit Vector 2:", u2)

print("Dot Product:", dot_product)
```
