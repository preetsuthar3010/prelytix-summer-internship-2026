## task 2

```python
countries = ['India', 'China', 'USA', 'Indonesia', 'Brazil']
population = [1420, 1410, 340, 280, 215]  # in millions (approx)

# Create horizontal bar chart
plt.barh(countries, population, color='lightgreen')

# Add labels and title
plt.xlabel('Population (millions)')
plt.title('Population of 5 Countries')

# Show the chart
```

```python
ages = np.random.randint(0, 81, 500)

# Plot histogram with 10 bins and black edges
plt.hist(ages, bins=10, edgecolor='black', color='skyblue')

# Add labels and title
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Histogram of Random Ages')

# Show the plot
plt.show()
```

```python
activities = ['Sleep', 'Work', 'Study', 'Leisure']
time_spent = [8, 9, 4, 3]  # hours per day

# Explode the 'Study' slice (index 2)
explode = (0, 0, 0.1, 0)

# Create pie chart
plt.pie(time_spent, labels=activities, explode=explode, autopct='%1.1f%%',
        colors=['lightblue', 'lightgreen', 'orange', 'pink'], startangle=90)

# Add title
plt.title('Daily Time Allocation')

# Show chart
plt.show()
```

## task 3

```python
x = np.linspace(0, 12, 400)  # restrict x from 0 to 12

# Define functions
y1 = x**2
y2 = x**3

# Plot both lines
plt.plot(x, y1, label='y = x^2', color='blue')
plt.plot(x, y2, label='y = x^3', color='red')

# Add title and labels
plt.title('Comparison of y = x^2 and y = x^3')
plt.xlabel('x-axis')
plt.ylabel('y-axis')

# Add grid
plt.grid(True)

# Set x-axis limits
plt.xlim(0, 12)

# Add legend
plt.legend()

# Show plot
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

scatter_x = np.random.rand(50)
scatter_y = np.random.rand(50)

bar_categories = ['A', 'B', 'C', 'D']
bar_values = [5, 7, 3, 8]

hist_data = np.random.randn(200)

# Create 2x2 subplot grid
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Line plot
axs[0, 0].plot(x, y, color='blue')
axs[0, 0].set_title('Line Plot')

# Scatter plot
axs[0, 1].scatter(scatter_x, scatter_y, color='red', marker='^')
axs[0, 1].set_title('Scatter Plot')

# Bar chart
axs[1, 0].bar(bar_categories, bar_values, color='green')
axs[1, 0].set_title('Bar Chart')

# Histogram
axs[1, 1].hist(hist_data, bins=10, edgecolor='black', color='orange')
axs[1, 1].set_title('Histogram')

# Adjust layout
plt.tight_layout()
plt.show()
```
