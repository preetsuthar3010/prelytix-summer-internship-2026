## Series Task 1

```
import pandas as pd

temps = pd.Series(
    [32.5, 35.1, 33.8, 31.2, 36.0, 34.5, 30.9],
    index=['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
)

print(temps)
print("Hottest Day:", temps.idxmax())
print("Coldest Day:", temps.idxmin())
```

## Series Task 2

```
import pandas as pd

s1 = pd.Series([10,20,30], index=['a','b','c'])
s2 = pd.Series([1,2,3], index=['b','c','d'])

print(s1 + s2)
print(s1.add(s2, fill_value=0))
```

## DataFrame Task

```
import pandas as pd

df = pd.DataFrame({
    'Name':['Alice','Bob','Carol','David','Eva'],
    'Math':[80,75,90,85,88],
    'Science':[78,82,89,80,91],
    'English':[85,79,88,84,90]
})

df['Total'] = df[['Math','Science','English']].sum(axis=1)

print(df)
```

## Titanic Filtering

```
import pandas as pd
import seaborn as sns

df = sns.load_dataset('titanic')

filtered = df[
    (df['sex']=='female') &
    (df['pclass']==1) &
    (df['survived']==1)
]

print(filtered.head())
print("Count:", len(filtered))
```
