# Pandas Day 4 Tasks Solutions

## Sorting Task

```
import pandas as pd
import seaborn as sns

df = sns.load_dataset('titanic')

top10 = df.sort_values('fare', ascending=False)[
    ['who','fare','pclass']
].head(10)

print(top10)
```

## Ranking Task

```
import pandas as pd

df = pd.DataFrame({
    'Name':['A','B','C','D','E'],
    'Marks':[90,85,90,70,85]
})

df['Rank'] = df['Marks'].rank(
    ascending=False,
    method='dense'
)

print(df)
```

## GroupBy Task

```
import pandas as pd
import seaborn as sns

df = sns.load_dataset('titanic')

print(df.groupby('sex')['survived'].mean())
print(df.groupby('pclass')['survived'].mean())
```

## Merge Task

```
import pandas as pd

students = pd.DataFrame({
    'id':[1,2,3],
    'name':['Alice','Bob','Carol']
})

grades = pd.DataFrame({
    'id':[1,2,4],
    'marks':[88,75,92]
})

print(pd.merge(students, grades, on='id', how='outer'))
```

## Pivot Table Task

```
import pandas as pd
import seaborn as sns

df = sns.load_dataset('titanic')

pivot = pd.pivot_table(
    df,
    values='fare',
    index='pclass',
    columns='sex',
    aggfunc='mean'
)

print(pivot)
```
