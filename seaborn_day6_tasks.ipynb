## task 1

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
penguins = sns.load_dataset("penguins")
sns.scatterplot(
    data=penguins,
    x='bill_length_mm',
    y='bill_depth_mm',
    hue='species',
    style='species',
)
plt.title('Penguins vs Bill_Length_Mm')
plt.show()
```

```python
flights = sns.load_dataset("flights")
sns.lineplot(
    data=flights,
    x='year',
    y='passengers',
    hue='month',
    style='month',
)
plt.title('Flights vs No_of_Passengers')
plt.show()
```

## task 3

```python
titanic = sns.load_dataset("titanic")
sns.histplot(
    data=titanic,
    x='age',
    hue = 'survived',
    kde=True,
    multiple='stack',
    bins=30,
)
plt.title('passenger survival by ages')
plt.xlabel('age')
plt.ylabel('survived')
plt.show()
```

```python
sns.countplot(
    data=titanic,
    x="pclass",       # passenger class (1, 2, 3)
    palette="Set2"    # optional: nice color palette
)

plt.title("Number of Passengers by Class on Titanic")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()
```

```python
sns.boxplot(
    data=titanic,
    x="class",
    y="survived",
)
plt.title("Survived by Class on Titanic")
plt.xlabel("Class")
plt.ylabel("Survived")
plt.show()
```

```python
tips = sns.load_dataset("tips")
corr = tips.corr(numeric_only=True)
print(corr)
sns.heatmap(
    corr,
    annot = True,
    cmap = "viridis",
    fmt=".2f"
)
plt.title("Correlation Matrix")
plt.show()
```
