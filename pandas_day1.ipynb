```python
import seaborn as sns

# Load Titanic dataset
titanic = sns.load_dataset("titanic")

# Print shape
print("Shape of dataset:")
print(titanic.shape)

# Print data types
print("\nData Types:")
print(titanic.dtypes)

# Print missing values count
print("\nMissing Values Per Column:")
print(titanic.isnull().sum())
```

task 3

```python
# Summary statistics
desc = titanic.describe()

print("Summary Statistics:")
print(desc)

# Column with highest standard deviation
highest_std = desc.loc['std'].idxmax()

print("\nColumn with Highest Standard Deviation:")
print(highest_std)

# Skewness of numeric columns
skewness = titanic.skew(numeric_only=True)

print("\nSkewness of Numeric Columns:")
print(skewness)

# Most skewed column
most_skewed = skewness.abs().idxmax()

print("\nMost Skewed Column:")
print(most_skewed)
```
